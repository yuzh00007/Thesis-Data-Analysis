```python
import base64

def base64_to_wav(base64_string, output_file):
    try:
        wav_data = base64.b64decode(base64_string)
        with open(output_file, 'wb') as f:
            f.write(wav_data)
        print(f'Converted base64 to {output_file} successfully!')
    except Exception as e:
        print(f'Error converting base64 to .wav: {e}')
```

```python
df = pd.read_csv("./dataRaw/experiment_6323_results.csv")
testworkers = ["test", "123", "testFirefox", "testRedirect", "testFinal"]  # add more to this list if there were more test runs

df = df[~df["workerid"].isin(testworkers)]
len(df["workerid"].unique())
```

```python
import os
import re
import json


MAIN_FOLDER_NAME = "./dataOutput/decodedAudio"

try:
    os.mkdir(MAIN_FOLDER_NAME)
except FileExistsError:
    ...

ignored_workers = ['69cc1fac2b52b1c4dd21cc7d']
workerid_list = [x for x in list(df["workerid"].unique()) if  x not in ignored_workers]

for worker in workerid_list:
    print(worker)

    # make folder
    worker_audio_folder = f"{MAIN_FOLDER_NAME}/{worker}_audio"
    try:
        os.mkdir(worker_audio_folder)
    except FileExistsError:
        ...


    # put audio streams into a list
    answers = df[df["workerid"] == worker]["answer"]
    audio_response = None
    for answer in answers:
        if (len(answer) > 3000):
            audio_response = answer

    results = json.loads(json.loads(audio_response))

    audio_streams = results["practiceTrials"]
    audio_streams.append(results["strategyFeedback"])

    # save audios
    for num, string in enumerate(audio_streams):
        first_comma_index = string.find(",") + 1
        base64_string = string.strip().replace("\"","")[first_comma_index:]
        file_name = f"{worker_audio_folder}/audio-{num + 1}.wav"
        base64_to_wav(base64_string, file_name)
```
